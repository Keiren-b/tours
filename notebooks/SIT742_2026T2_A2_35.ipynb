#### Optional month-label helper example

This optional helper example only shows one way to convert labels such as `2023M03` into a month-aware representation. You may implement your own helper inside Q1. Do not use string subtraction for month arithmetic.


````python
def month_label_to_period_example(month_label):
    """Optional example: convert a YYYYMmm label to a pandas monthly Period."""
    return pd.Period(str(month_label).replace("M", "-"), freq="M")


def period_to_month_label_example(period):
    """Optional example: convert a pandas monthly Period back to YYYYMmm."""
    return f"{period.year}M{period.month:02d}"
````

````python
def generate_naive_forecast_wide(
    historical_actual_wide,
    cutoff_label,
    forecast_months,
    required_destinations=None,
    lag=1,
    date_column="Date",
):
    """Return the required wide-format naive-lag forecast DataFrame.

    Return exactly one row per requested month in the supplied forecast_months
    order. Return columns as date_column followed by the inferred or supplied
    destination order, with no extra columns. If required_destinations is None,
    use all columns except date_column in historical_actual_wide.

    For each forecast month m, use m - lag months as the source month. Use
    observed values at or before cutoff_label; otherwise use previously generated
    forecasts recursively when available. Raise a clear error if lag is not a
    positive integer. Use month-aware labels, not string subtraction.
    """
    ##
    naive_forecast_wide = []

    # -----------------------------------------
    # Input Checks
    # -----------------------------------------

    #check lag is positive integer
    if not isinstance(lag, int) or lag < 1:
      raise ValueError(f'Lag value entered is {lag} with type {type(lag)}. Lag must be a positive integer')

    #sort dataframe by date
    historical_data = historical_actual_wide.sort_values(by=date_column)

    # Handle required_destinations input, checking that all destinations are actually in the dataset
    if required_destinations is None:
      required_destinations = [column for column in historical_actual_wide.columns if column != date_column]
    else:
      required_destinations = list(required_destinations)
      missing = [column for column in required_destinations if column not in historical_actual_wide.columns]
      if missing:
        raise ValueError(f"These destinations are not in the dataset: {missing}")

    # ----------------------------------------------------------------------------------
    # Build a table of values that includes only the destinations and the date specified
    # AUse this table to find the lag values, and append forecasts to it so later searches can find these values
    # ----------------------------------------------------------------------------------

    # build df with date column first, and then the destinations in the order they appear or the order they are inputted

    df_cut = historical_data.copy()
    df_cut = df_cut[
        [date_column] + required_destinations
    ]

    # remove rows after cuttoff date
    if cutoff_label is not None:
      cut_date = month_label_to_period(cutoff_label)
      dates_in_window = df_cut[date_column].map(month_label_to_period) <= cut_date
      df_cut = df_cut[dates_in_window]

    # check that cuttoff was applied correctly
      latest_month = df_cut[date_column].map(month_label_to_period).max()
      assert latest_month <= cut_date

    # make date column an index for easier lookup
    df_cut = df_cut.set_index(date_column)



    # ----------------------------------------------------------------------------------
    # Build the forecast table
    # ----------------------------------------------------------------------------------


    # loop through forcast dates, calculate lag month and lookup
    x = {}
    sorted_months = [month_label_to_period(date) for date in forecast_months]
    sorted_months.sort()
    sorted_months = [period_to_month_label_example(date) for date in sorted_months]

    for forecast_date in sorted_months:
      lag_date = month_label_to_period(forecast_date) - lag
      lag_date = period_to_month_label_example(lag_date)

    # checkif lag date exists in data
      try:
        lag_row = df_cut.loc[lag_date]

      except KeyError:
        raise ValueError(f'The Lag date {lag_date} is not in data')

    # add the forecast value to the table for future lookup
      df_cut.loc[forecast_date] = lag_row

    # build the row to append
      forecast_row = {date_column: forecast_date}
      for destination in required_destinations:
        forecast_row[destination] = lag_row[destination]

      x[forecast_date] = forecast_row


    for date in forecast_months:
      naive_forecast_wide.append(x[date])

    naive_forecast_wide = pd.DataFrame(naive_forecast_wide)
    return naive_forecast_wide

````

`df` means a pandas DataFrame. Both inputs use a wide table: one row per month, one `Date` column, and one numeric column per destination. This is not a long table.

Modes: forecast-only mode (`actual_wide_df=None`) checks export readiness; forecast-vs-actual mode checks whether forecast and actual/result tables can align by `Date` and destination.

Requirements: check required `Date`, months, destination columns, row counts, duplicate dates, extra rows/columns, missing values, numeric values, finite values, and alignment when actuals are supplied. If `required_destinations=None`, infer destination columns from `forecast_wide_df`, but all-20 evidence must still match `REQUIRED_DESTINATIONS`.

Return a Python dictionary containing every key in `Q2_AUDIT_FIELDS`, in that order. In forecast-only mode, `can_align` and all actual-related fields must be `None`, not omitted. In forecast-vs-actual mode, `actual_row_count` counts actual rows after subsetting to `required_months`; `can_align` reports structural alignment; `is_valid` reports all applicable structure and value-quality checks. Duplicate counts count occurrences after the first occurrence.

````python
def validate_forecast_actual_wide(
    forecast_wide_df,
    actual_wide_df=None,
    required_months=None,
    required_destinations=None,
    date_column="Date",
):
    """Return the fixed Q2 audit dictionary for wide-table validation.

    Populate every key in Q2_AUDIT_FIELDS. Forecast-only mode uses None for
    can_align and every actual-related field. Forecast-vs-actual mode subsets
    actual_wide_df to required_months before reporting actual_row_count and
    checking structural alignment.
    """
    audit = {
        "is_valid": False,
        "can_align": None if actual_wide_df is None else False,
        "forecast_row_count": len(forecast_wide_df),
        "actual_row_count": None,
        "expected_row_count": (
            len(required_months) if required_months is not None else None
        ),
        "missing_months_in_forecast": [],
        "missing_months_in_actual": None,
        "extra_months_in_forecast": [],
        "extra_months_in_actual_source": None,
        "missing_destinations_in_forecast": [],
        "missing_destinations_in_actual": None,
        "extra_columns_in_forecast": [],
        "duplicate_forecast_dates": 0,
        "duplicate_actual_dates": None,
        "missing_forecast_value_count": 0,
        "missing_actual_value_count": None,
        "nonnumeric_forecast_value_count": 0,
        "nonnumeric_actual_value_count": None,
        "nonfinite_forecast_value_count": 0,
        "nonfinite_actual_value_count": None,
    }

    # --------------------------
    # Infer destination columns and months
    # --------------------------

    if required_destinations is None:
      required_destinations = [column for column in forecast_wide_df.columns if column != date_column]

    if required_months is None:
      required_months = [column for column in forecast_wide_df[date_column]]



    # --------------------------
    # row counts
    # --------------------------

    if actual_wide_df is not None:

      required_mask = actual_wide_df[date_column].isin(required_months)
      audit["actual_row_count"] = len(actual_wide_df[required_mask])

    # --------------------------
    # Missing months
    # --------------------------
    audit["missing_months_in_forecast"] = [missing for missing in required_months
                                            if missing not in forecast_wide_df[date_column].to_list()]

    if actual_wide_df is not None:
      audit["missing_months_in_actual"] = [missing for missing in required_months
                                            if missing not in actual_wide_df[date_column].to_list()]

    # --------------------------
    # Extra months
    # --------------------------
    audit["extra_months_in_forecast"] = [extra for extra in forecast_wide_df[date_column]
                                          if extra not in required_months]
    if actual_wide_df is not None:
      audit["extra_months_in_actual_source"] = [extra for extra in actual_wide_df[date_column]
                                          if extra not in required_months]

    # --------------------------
    # Missing destinations
    # --------------------------

    forecast_destinations = forecast_wide_df.columns.drop(date_column)
    audit["missing_destinations_in_forecast"] = [destination for destination in required_destinations
                                                  if destination not in forecast_destinations]

    if actual_wide_df is not None:
      actual_destinations = actual_wide_df.columns.drop(date_column)
      audit["missing_destinations_in_actual"] = [destination for destination in required_destinations
                                                  if destination not in actual_destinations]
    # --------------------------
    # Extra columns in forecast
    # --------------------------

    audit["extra_columns_in_forecast"] = [extra for extra in forecast_wide_df.columns.drop(date_column)
                                          if extra not in required_destinations]


    # --------------------------
    # Duplicates, Missing, Inf
    # --------------------------

    audit["duplicate_forecast_dates"] = int(forecast_wide_df.duplicated(subset=[date_column]).sum())
    if actual_wide_df is not None:
      audit["duplicate_actual_dates"] = int(actual_wide_df.duplicated(subset=[date_column]).sum())

    # --------------------------
    # Missing Values
    # --------------------------

    drop_cols = [extra for extra in forecast_wide_df.columns.drop(date_column)
                  if extra not in required_destinations]

    forecast_cut = forecast_wide_df.drop(columns=drop_cols)
    forecast_cut = forecast_cut[forecast_cut[date_column].isin(required_months)]
    forecast_cut = forecast_cut.drop(columns=[date_column])


    audit["missing_forecast_value_count"] = int(forecast_cut.isna().sum().sum())

    if actual_wide_df is not None:
      drop_cols = [extra for extra in forecast_wide_df.columns.drop(date_column)
        if extra not in required_destinations]
      actual_cut = actual_wide_df.drop(columns=drop_cols)
      actual_cut = actual_cut[actual_cut[date_column].isin(required_months)]
      actual_cut = actual_cut.drop(columns=[date_column])
      audit["missing_actual_value_count"] = int(actual_cut.isna().sum().sum())

    # --------------------------
    # Non-numeric Values
    # --------------------------

    audit["nonnumeric_forecast_value_count"] = int(forecast_cut.apply(
        lambda x: (x.notnull() & pd.to_numeric(x, errors='coerce')
        .isnull())
        .sum()).sum())

    if actual_wide_df is not None:
        audit["nonnumeric_actual_value_count"] = int(actual_cut.apply(
        lambda x: (x.notnull() & pd.to_numeric(x, errors='coerce')
        .isnull())
        .sum()).sum())

    # --------------------------
    # Non-finite value
    # --------------------------

    forecast_numeric = forecast_cut.apply(pd.to_numeric, errors="coerce")
    audit["nonfinite_forecast_value_count"] = int(np.isinf(forecast_numeric).sum().sum())

    if actual_wide_df is not None:
      actual_numeric = actual_cut.apply(pd.to_numeric, errors="coerce")
      audit["nonfinite_actual_value_count"] = int(np.isinf(actual_numeric).sum().sum())


    # --------------------------
    # can align
    # --------------------------


    if actual_wide_df is None:
      audit["can_align"] = None
    else:
      audit["can_align"] = all([
          audit["missing_months_in_forecast"] == [],
          audit["missing_months_in_actual"] == [],
          audit["missing_destinations_in_forecast"] == [],
          audit["missing_destinations_in_actual"] == [],
          audit["duplicate_forecast_dates"] == 0,
          audit["duplicate_actual_dates"] == 0
          ]
      )

    # --------------------------
    # is valid
    # --------------------------

    audit["is_valid"] = all([
        date_column in forecast_wide_df.columns,
        audit["can_align"] is not False,
        audit["duplicate_forecast_dates"] == 0,
        audit["actual_row_count"] == audit["expected_row_count"] if actual_wide_df is not None else True,
        audit["missing_months_in_forecast"] == [],
        audit["missing_months_in_actual"] == [] if actual_wide_df is not None else True,
        audit["extra_months_in_forecast"] == [],
        audit["missing_destinations_in_forecast"] == [],
        audit["extra_columns_in_forecast"] == [],
        audit["missing_forecast_value_count"] == 0,
        audit["missing_actual_value_count"] == 0 if actual_wide_df is not None else True,
        audit["nonnumeric_forecast_value_count"] == 0,
        audit["nonnumeric_actual_value_count"] == 0 if actual_wide_df is not None else True,
        audit["nonfinite_forecast_value_count"] == 0,
        audit["nonfinite_actual_value_count"] == 0 if actual_wide_df is not None else True,
    ])


# The audit should show is_valid=True, correct row counts, no missing months, no missing destinations, no duplicate dates, and numeric finite
# values.
    return audit
````

### Q3. Forecast accuracy measures from forecast and actual tables [5 marks]

Implement `evaluate_forecast_wide(forecast_wide_df, actual_wide_df, training_actual_wide_df, start_month, end_month, required_destinations=None, naive_lag=1, date_column="Date")`.

Use the [A2 Forecast Measures Guide](SIT742-2026T2-A2-Forecast-Measures-Guide.md). Both MASE and MAPE are required.

Requirements: align forecast and actual/result tables by `Date` and destination; work destination by destination; select months from `start_month` to `end_month`; evaluate all forecast destination columns when `required_destinations=None`; compute MAE, MASE, MAPE, denominator, `n`, `n_denominator_pairs`, `mase_available`, and `denominator_warning`; keep MASE/MAPE separately identifiable; sort training actuals chronologically; use all valid lagged training pairs for each destination's MASE denominator; use `naive_lag=1` for official validation unless stated otherwise; handle missing values, nonfinite values, zero actuals for MAPE, and unavailable denominators explicitly.

Return exactly `(destination_metrics, aggregate_metrics)`. `destination_metrics` must be a pandas DataFrame with columns in `DESTINATION_METRIC_COLUMNS`. `aggregate_metrics` must be a Python dictionary with exactly the keys in `AGGREGATE_METRIC_FIELDS`.

Do not fill missing denominators with zero. Do not use validation actuals or assessment-period actual/result values to repair denominators. `lag=12` baseline forecasts are for comparison; they are not the official MASE denominator. Aggregate measures are unweighted summaries of destination-level measures unless otherwise stated.

**Required answer markers:** Do not delete, rename, edit, move, or reorder the Q3 START/END marker comments below. Write your response between them.

<!-- A2:ANSWER:Q3:START -->
Write any brief Q3 performance-measure design notes here if needed. Delete this sentence before submission.

````python
def evaluate_forecast_wide(
    forecast_wide_df,
    actual_wide_df,
    training_actual_wide_df,
    start_month,
    end_month,
    required_destinations=None,
    naive_lag=1,
    date_column="Date",
):
    """Return (destination_metrics, aggregate_metrics).

    destination_metrics is a DataFrame using DESTINATION_METRIC_COLUMNS.
    aggregate_metrics is a dictionary using AGGREGATE_METRIC_FIELDS.
    Sort training_actual_wide_df chronologically before computing all valid
    denominator pairs. If required_destinations is None, evaluate all forecast
    columns except date_column; otherwise evaluate exactly those destinations.
    """
    destination_metrics = pd.DataFrame(columns=DESTINATION_METRIC_COLUMNS)
    aggregate_metrics = {
        field: None for field in AGGREGATE_METRIC_FIELDS
    }



    # -----------------------------
    # destinations
    # -----------------------------

    if required_destinations is None:
        destinations = [c for c in forecast_wide_df.columns if c != date_column]
    else:
        destinations = list(required_destinations)

    # -----------------------------
    # evaluation month range
    # -----------------------------

    start_period = month_label_to_period(start_month)
    end_period = month_label_to_period(end_month)

    # -----------------------------
    # select forecast and actual
    # -----------------------------

    forecast = forecast_wide_df[[date_column] + destinations].copy()
    actual = actual_wide_df[[date_column] + destinations].copy()

    forecast["_period"] = forecast[date_column].map(month_label_to_period)
    actual["_period"] = actual[date_column].map(month_label_to_period)

    forecast = forecast[forecast["_period"].between(start_period, end_period)]
    actual = actual[actual["_period"].between(start_period, end_period)]

    # sort chronologically
    forecast = forecast.sort_values("_period")
    actual = actual.sort_values("_period")

    # align forecast and actual by Date
    forecast = forecast.set_index(date_column)
    actual = actual.set_index(date_column)

    common_dates = forecast.index.intersection(actual.index)

    forecast = forecast.loc[common_dates]
    actual = actual.loc[common_dates]

    # -----------------------------
    # sort training history
    # -----------------------------

    training = training_actual_wide_df[[date_column] + destinations].copy()
    training["_period"] = training[date_column].map(month_label_to_period)
    training = training.sort_values("_period")

    # -----------------------------
    # calculate destination metrics
    # -----------------------------

    rows = []

    for destination in destinations:

        f = pd.to_numeric(forecast[destination], errors="coerce")
        a = pd.to_numeric(actual[destination], errors="coerce")

        # valid forecast/actual pairs
        valid = f.notna() & a.notna() & np.isfinite(f) & np.isfinite(a)

        f_valid = f[valid]
        a_valid = a[valid]
        n = len(f_valid)

        # -------------------------
        # MAE
        # -------------------------

        mae = float(np.abs(a_valid - f_valid).mean()) if n else np.nan

        # -------------------------
        # MAPE
        # -------------------------
        # Zero actual values are excluded.

        mape_valid = valid & a.ne(0)

        if mape_valid.any():
            mape = float(
                (np.abs((a[mape_valid] - f[mape_valid]) / a[mape_valid]) * 100).mean()
            )
        else:
            mape = np.nan

        # -------------------------
        # MASE denominator
        # -------------------------
        # use all valid lagged training pairs.

        training_values = pd.to_numeric(
            training[destination], errors="coerce"
        ).to_numpy(dtype=float)

        if len(training_values) > naive_lag:
            current = training_values[naive_lag:]
            previous = training_values[:-naive_lag]

            valid_pairs = np.isfinite(current) & np.isfinite(previous)
            denominator_errors = np.abs(current[valid_pairs] - previous[valid_pairs])
        else:
            denominator_errors = np.array([], dtype=float)

        n_denominator_pairs = len(denominator_errors)

        denominator = (
            float(denominator_errors.mean())
            if n_denominator_pairs
            else np.nan
        )

        # -------------------------
        # MASE
        # -------------------------

        mase_available = (
            n > 0
            and np.isfinite(mae)
            and n_denominator_pairs > 0
            and np.isfinite(denominator)
            and denominator > 0
        )

        if mase_available:
            mase = float(mae / denominator)
            denominator_warning = False
        else:
            mase = np.nan
            denominator_warning = True

        # -------------------------
        # store destination metrics
        # -------------------------

        rows.append({
            "destination": destination,
            "n": int(n),
            "mae": mae,
            "mase": mase,
            "mape": mape,
            "denominator": denominator,
            "n_denominator_pairs": int(n_denominator_pairs),
            "mase_available": bool(mase_available),
            "denominator_warning": bool(denominator_warning)
        })

    # -----------------------------
    # destination metrics
    # -----------------------------

    destination_metrics = pd.DataFrame(
        rows, columns=DESTINATION_METRIC_COLUMNS
    )

    # -----------------------------
    # aggregate metrics
    # -----------------------------

    mase_values = destination_metrics["mase"].dropna()
    mape_values = destination_metrics["mape"].dropna()

    aggregate_metrics = {
        "n_destinations": len(destination_metrics),
        "mean_mase": float(mase_values.mean()) if len(mase_values) else np.nan,
        "median_mase": float(mase_values.median()) if len(mase_values) else np.nan,
        "mean_mape": float(mape_values.mean()) if len(mape_values) else np.nan
    }

    return destination_metrics, aggregate_metrics
````

We reshaped the tourism data into a long format table that contains `Date`, `Destination` and `Demand`. We restricted the modelling data training cutoff to 2023M02 to prevent future information leakage.

Naive lag-1, seasonal Naive lag-12 and ARIMA models were comapred using the required validation period of 2023M03 to 2023M07 across all 20 destinations. With the ARIMA model, grid search was conducted to assess whether alternative specifications could improve validation performance. This was compared using MASE and MAPE.

Validation performance of the 3 models: Naive lag-1, Naive lag-12 and ARIMA were assesed using MAE, MASE and MAPE. The ARIMA model was selected because it achieved the lowest validation mean MASE with mean MAPE used as a secondary comparison criteria.

````python

# A zero-row table with these columns is valid only when no external data is used.
external_feature_log = pd.DataFrame(columns=EXTERNAL_FEATURE_LOG_COLUMNS)

# ------------------------------------------------------------
#  prepare data in long format for the models
# ------------------------------------------------------------
tourism_series_long = raw_tourism_data.melt(id_vars="Date", var_name="destination", value_name="demand")
tourism_series_long = tourism_series_long.sort_values(["destination", "Date"]).reset_index(drop=True)

# ------------------------------------------------------------
# Prepare cutoff-safe modelling data
# ------------------------------------------------------------
modelling_data_or_feature_table = raw_tourism_data[raw_tourism_data["Date"].map(month_label_to_period) <= month_label_to_period(TRAIN_END)].sort_values("Date").reset_index(drop=True)

destinations = [c for c in modelling_data_or_feature_table.columns if c != FINAL_FORECAST_DATE_COLUMN]

# ------------------------------------------------------------
# Generate naive validation forecasts
# ------------------------------------------------------------
baseline_lag1_forecast = generate_naive_forecast_wide(modelling_data_or_feature_table, TRAIN_END, VALIDATION_MONTHS, lag=1)
baseline_lag12_forecast = generate_naive_forecast_wide(modelling_data_or_feature_table, TRAIN_END, VALIDATION_MONTHS, lag=12)

#------------------------------------------------------------
# grid search to select best ARIMA model
#------------------------------------------------------------
RUN_GRID_SEARCH = True
candidate_orders = [
    (p, d, q)
    for p in range(0, 3)
    for d in range(0, 2)
    for q in range(0, 3)]

grid_results = []
for order in candidate_orders:
    try:
        candidate_rows = []

        for destination in destinations:
            history = modelling_data_or_feature_table[destination].dropna().astype(float).tolist()

            for month in VALIDATION_MONTHS:
                model = ARIMA(history, order=order)
                model_fit = model.fit()
                forecast = float(model_fit.forecast(steps=1)[0])

                actual = raw_tourism_data.loc[raw_tourism_data["Date"] == month, destination].iloc[0]

                candidate_rows.append({
                    "Date": month,
                    "destination": destination,
                    "forecast": forecast})

                history.append(float(actual))

        candidate_forecast = pd.DataFrame(candidate_rows).pivot(
            index="Date",
            columns="destination",
            values="forecast"
        ).reset_index()

        _, candidate_aggregate = evaluate_forecast_wide(
            candidate_forecast,
            validation_actual_wide_2023M03_2023M07,
            modelling_data_or_feature_table,
            VALIDATION_START,
            VALIDATION_END,
            naive_lag=1 )

        grid_results.append({
            "order": order,
            "mean_mase": candidate_aggregate["mean_mase"],
            "median_mase": candidate_aggregate["median_mase"],
            "mean_mape": candidate_aggregate["mean_mape"]})

    except Exception as exc:
        grid_results.append({
            "order": order,
            "mean_mase": np.nan,
            "median_mase": np.nan,
            "mean_mape": np.nan,
            "error": str(exc)})

grid_results = pd.DataFrame(grid_results)
grid_results = grid_results.sort_values(["mean_mase", "mean_mape"], na_position="last").reset_index(drop=True)
display(grid_results)

# Use the best ARIMA order found by grid search
ARIMA_ORDER = tuple(grid_results.iloc[0]["order"])
print("Selected ARIMA order from grid search:", ARIMA_ORDER)
print("Grid-search mean MASE:", grid_results.iloc[0]["mean_mase"])
print("Grid-search mean MAPE:", grid_results.iloc[0]["mean_mape"])

# ------------------------------------------------------------
# Generate ARIMA validation forecasts
# ------------------------------------------------------------

arima_validation_rows = []
for destination in destinations:
    history = modelling_data_or_feature_table[destination].dropna().astype(float).tolist()

    for month in VALIDATION_MONTHS:
        model = ARIMA(history, order=ARIMA_ORDER)
        model_fit = model.fit()

        forecast = float(model_fit.forecast(steps=1)[0])

        arima_validation_rows.append({
            "model_label": "arima",
            "Date": month,
            "destination": destination,
            "forecast": forecast
        })

        actual = raw_tourism_data.loc[ raw_tourism_data["Date"] == month, destination ].iloc[0]

        history.append(float(actual))

arima_validation_forecast = pd.DataFrame( arima_validation_rows, columns=VALIDATION_FORECAST_COLUMNS)

# ------------------------------------------------------------
# combine validation forecasts
# ------------------------------------------------------------
validation_forecasts = pd.concat(
    [
        baseline_lag1_forecast.melt(
            id_vars=["Date"],
            var_name="destination",
            value_name="forecast"
        ).assign(model_label="naive_lag1"),

        baseline_lag12_forecast.melt(
            id_vars=["Date"],
            var_name="destination",
            value_name="forecast"
        ).assign(model_label="naive_lag12"),

        arima_validation_forecast
    ], ignore_index=True)

validation_forecasts = validation_forecasts[ VALIDATION_FORECAST_COLUMNS]
validation_forecasts["forecast"] = pd.to_numeric(validation_forecasts["forecast"], errors="coerce")

# ------------------------------------------------------------
# Evaluate naive lag-1
# ------------------------------------------------------------
lag1_metrics, lag1_aggregate = evaluate_forecast_wide(
    baseline_lag1_forecast,
    validation_actual_wide_2023M03_2023M07,
    modelling_data_or_feature_table,
    VALIDATION_START,
    VALIDATION_END,
    naive_lag=1)

# ------------------------------------------------------------
# Evaluate naive lag-12
# ------------------------------------------------------------
lag12_metrics, lag12_aggregate = evaluate_forecast_wide(
    baseline_lag12_forecast,
    validation_actual_wide_2023M03_2023M07,
    modelling_data_or_feature_table,
    VALIDATION_START,
    VALIDATION_END,
    naive_lag=1)

# ------------------------------------------------------------
# Evaluate ARIMA
# ------------------------------------------------------------
arima_validation_forecast_wide = arima_validation_forecast.pivot( index="Date", columns="destination", values="forecast").reset_index()

arima_metrics, arima_aggregate = evaluate_forecast_wide(
    arima_validation_forecast_wide,
    validation_actual_wide_2023M03_2023M07,
    modelling_data_or_feature_table,
    VALIDATION_START,
    VALIDATION_END,
    naive_lag=1)

# ------------------------------------------------------------
# Combine validation metrics
# ------------------------------------------------------------
validation_metrics = pd.concat(
    [lag1_metrics.assign(model_label="naive_lag1"),
        lag12_metrics.assign(model_label="naive_lag12"),
        arima_metrics.assign(model_label="arima")], ignore_index=True)

validation_metrics = validation_metrics[ VALIDATION_METRIC_COLUMNS]

# ------------------------------------------------------------
# Compare models and select best model
# ------------------------------------------------------------
model_results = pd.DataFrame([
    {
        "model_label": "naive_lag1",
        "model_family": "Naive",
        **lag1_aggregate
    },
    {
        "model_label": "naive_lag12",
        "model_family": "Seasonal Naive",
        **lag12_aggregate
    },
    {
        "model_label": "arima",
        "model_family": "ARIMA",
        **arima_aggregate
    }
])

# Select the model with the lowest validation mean MASE.
selected_model = (
    model_results
    .sort_values(
        ["mean_mase", "mean_mape"],
        na_position="last"
    )
    .iloc[0]["model_label"]
)

model_summary = model_results.copy()
model_summary["selected"] = (model_summary["model_label"] == selected_model)
model_summary["selection_rationale"] = model_summary[ "model_label"].apply( lambda x: "Selected because it has the lowest validation mean MASE."
    if x == selected_model
    else "Not selected because another model performed better on validation mean MASE.")
model_summary["reproducibility_notes"] = (f"Deterministic model fitting; ARIMA order={ARIMA_ORDER}; ""training cutoff=2023M02; validation=2023M03 to 2023M07.")
model_summary["stability_notes"] = ("Final forecasts checked for missing, non-finite and negative values.")
model_summary = model_summary[MODEL_SUMMARY_COLUMNS]

# ------------------------------------------------------------
# Create final forecast
# ------------------------------------------------------------
final_training_data = raw_tourism_data[raw_tourism_data["Date"].map(month_label_to_period)<= month_label_to_period(VALIDATION_END)].sort_values("Date").reset_index(drop=True)

if selected_model == "naive_lag1":
    forecast_submission_wide = generate_naive_forecast_wide(
        final_training_data,
        VALIDATION_END,
        FINAL_FORECAST_MONTHS,
        lag=1)

elif selected_model == "naive_lag12":
    forecast_submission_wide = generate_naive_forecast_wide(
        final_training_data,
        VALIDATION_END,
        FINAL_FORECAST_MONTHS,
        lag=12)

else:
    final_forecast_rows = []
    for destination in destinations:
        series = pd.to_numeric(final_training_data[destination], errors="coerce").dropna().astype(float).tolist()

        model = ARIMA(series, order=ARIMA_ORDER)
        model_fit = model.fit()

        forecast = model_fit.forecast(steps=len(FINAL_FORECAST_MONTHS))

        for month, value in zip(FINAL_FORECAST_MONTHS, forecast):
            final_forecast_rows.append({"Date": month, "destination": destination,"forecast": float(value)})

    forecast_submission_wide = pd.DataFrame(final_forecast_rows).pivot(index="Date",columns="destination",values="forecast").reset_index()

forecast_submission_wide = forecast_submission_wide[[FINAL_FORECAST_DATE_COLUMN] + destinations]

# ------------------------------------------------------------
# Final forecast stability checks
# ------------------------------------------------------------
forecast_submission_wide = forecast_submission_wide[[FINAL_FORECAST_DATE_COLUMN] + destinations].sort_values(FINAL_FORECAST_DATE_COLUMN).reset_index(drop=True)

forecast_values = forecast_submission_wide[destinations].apply(pd.to_numeric, errors="coerce")

assert len(forecast_submission_wide) == len(FINAL_FORECAST_MONTHS)
assert forecast_submission_wide[FINAL_FORECAST_DATE_COLUMN].tolist() == FINAL_FORECAST_MONTHS
assert forecast_values.notna().all().all()
assert np.isfinite(forecast_values.to_numpy()).all()
assert (forecast_values >= 0).all().all()

# ------------------------------------------------------------
# Print validation checks
# ------------------------------------------------------------
print("Selected model:", selected_model)
print("ARIMA order:", ARIMA_ORDER)

display(validation_metrics)
display(model_summary)
````

### Part A
MASE and MAPE provide two different views of forecast performance, hence both were used in analysis. MASE measures the forecast error relative to the typical movement in the training data, which makes it useful for comparing destinatons with different levels of demand. A lower MASE indicates that the forecast error is smaller relative to the historical movement of the destination.

It can be affected by the quality of its denominator, i.e. if there's few valid training pairs or historical movement is too small, the resulting MASE can be less representative. Therefore, the denominator and number of valid pairs were therefre shown in the validation_metrics results.

MAPE measures the forecast error as a percentage of actual demand, making it easier to interpret, however, it can become unstable when the actual demand is small and it can't be calculated when the actual value is zero. That's why MAPE was used as an additional measure, but not used instead of MASE.

The baseline_validation_summary shows that the lag-1 naive forecast performed better than the seasonal lag-12 baseline overall. The selected model evidence in model_summary shows that the ARIMA model improved compared to the baseline forecasts based on MASE. validation_metrics provides the destination-levl MASE and MAPE results, showing that the measures do not always give the same picture across destinations.
